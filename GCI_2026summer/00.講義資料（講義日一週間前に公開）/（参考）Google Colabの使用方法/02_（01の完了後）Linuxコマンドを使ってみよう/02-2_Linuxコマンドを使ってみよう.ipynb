### 3. GoogleドライブをColabにマウントする

Google Colabでは、自分のGoogleドライブをマウントして利用できます。これにより、ファイルの保存や操作が簡単になります。
ドライブのマウントの詳細については、「Google Colabの使用方法」の 5. ドライブマウントも参照してください。

#### 手順:
1. 次のコードをセルに入力して実行します。

````python
from google.colab import drive
drive.mount('/content/drive')
````

2. 実行すると、認証リンクが表示されます。リンクをクリックし、Googleアカウントでログインして、認証コードを取得してください。
3. 認証コードをColabのセルに貼り付けて、マウントを完了します。

#### マウント後のフォルダ構成図（個人の環境によって異なります）

```
/content/drive/
    └── MyDrive/
        ├── Colab Notebooks/
                   └── data/
  
```

- **MyDrive**: 自分のGoogleドライブのファイルやフォルダがここに表示されます。

#### 3.1 マウントされた自分のGoogleドライブのフォルダを閲覧する

````python
cd /content/drive/MyDrive/Colab Notebooks
````

````python
ls
````

#### 3.2 **新しいフォルダを作成する (`mkdir`)**

新しいフォルダを作成するには、`mkdir` コマンドを使います。

````python
mkdir data
````
